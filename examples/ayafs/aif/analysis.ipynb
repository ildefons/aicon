```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_delays(csv_path):
    df = pd.read_csv(csv_path)

    # Compute delay per message
    df['delay'] = df['time_out'] - df['time_in']

    plt.figure(figsize=(10, 6))
    plt.plot(df['id'], df['delay'], marker='o', linestyle='-', label='Delay (s)')
    plt.xlabel("Message ID")
    plt.ylabel("End-to-End Delay (s)")
    plt.title("End-to-End Delay per Message")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
plot_delays("/home/ildefons/yaf310/examples/aif/results/sim_trace.csv")
```
